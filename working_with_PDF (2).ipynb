```python
#read a PDF file using python
from PyPDF2 import PdfReader
reader = PdfReader("/content/jeff101.pdf")
print("Number of pages : ", len(reader.pages))
page = reader.pages[0]
text = page.extract_text()
print(text)
```

```python
#Read All Pages from a pdf
from PyPDF2 import PdfReader
reader = PdfReader("/content/jeff101.pdf")
for page_number,page in enumerate(reader.pages, start = 2):
  print(f"\n---page {page_number}---")
  text = page.extract_text()
  print(text)
```

```python
#read a specific page
from PyPDF2 import PdfReader
reader = PdfReader("/content/jeff101.pdf")
page = reader.pages[3]
print(page.extract_text())
```

```python
#search for a word in a PDF
from PyPDF2 import PdfReader
reader = PdfReader("/content/jeff101.pdf")
search_word = "God"
for page_number, page in enumerate(reader.pages, start=1):
  text = page.extract_text()
  if search_word in text:
    print(f"Found '{search_word}' on page {page_number}")
  else:
    print(f"Did not find '{search_word}' on page {page_number}")
```

```python
#count words in a pdf
from PyPDF2 import PdfReader
reader = PdfReader("/content/jeff101.pdf")
word_count = 0
for page in reader.pages:
  text = page.extract_text()
  if text:
    word_count += len(text.split())
print(f"Total word count in the PDF: {word_count}")
```

```python
#save extracted text into a .txt file
from PyPDF2 import PdfReader
reader = PdfReader("/content/jeff101.pdf")
with open("pdf_text.txt", "w", encoding="utf-8") as file:
  for page_number,page in enumerate (reader.pages,start=1):
    text = page.extract_text()

    file.write(f"\n---Page {page_number}---\n")
    file.write(text if text else "")
print("Text extracted and saved to 'extracted_text.txt'.")
```

```python
#copy the first page into another PDF
from PyPDF2 import PdfReader, PdfWriter
reader = PdfReader("/content/jeff101.pdf") # real original pdf
writer = PdfWriter() #create new pdf
writer.add_page(reader.pages[0]) #copy first page
#save as another PDF
with open("first_page.pdf", "wb") as output:
  writer.write(output)
print("First page copied to 'first_page.pdf'.")
```

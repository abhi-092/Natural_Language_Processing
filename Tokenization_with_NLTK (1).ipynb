```python
#10 count only words
import string
text = "python is easy and powerful"
tokens = word_tokenize(text)
words = [word for word in tokens if word not in string.punctuation]
print("Words:",words)
print("Tokens:",tokens)
print("numbers of tokens:", len(tokens))
```

```python
#13 tokenizing a paragraph
from nltk.tokenize import word_tokenize,sent_tokenize
text = """Python is easy to learn . NLTK provides many NLP tools. tokenization is an important NLP task"""
sentences = sent_tokenize(text)
print("Number of sentence: ", len(sentences))
for i , sentence in enumerate(sentences,start=1):
  print(f"n/Sentence {i} :",sentence)
  print(sentence)
  tokens = word_tokenize(sentence)
  print("Tokens:",tokens)
  print("Number of tokens:",len(tokens))
```

```python
#13 tokenizing a paragraph
from nltk.tokenize import sent_tokenize, word_tokenize
text="""
Artificila Intelegence is changing the world.
NLTK is a part of python.
Machine Learning is a subset of AI.
"""
sentences=sent_tokenize(text)
print("Number of scentences:",len(sentences))
for i, sentence in enumerate(sentences, start=1):
    print(f"\nSentence {i}:")
    print(sentence)

    tokens=word_tokenize(sentence)
    print("Tokens:",tokens)
```

```python
#15 complete tokenization program
import nltk
import string
from nltk.tokenize import sent_tokenize, word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')
text = """Python is easy to learn . NLTK provides many NLP tools. tokenization is an important NLP task"""
sentences = sent_tokenize(text)
print("sentences")
#step 1 sent tokenize
for sentence in sentences:
  print(sentence)

#step 2 : Word Tokenization
tokens = word_tokenize(text)
print("\nAll Tokens:")
print(tokens)

#step 3 : Remove Punctuation
words = [word for word in tokens if word not in string.punctuation]
print("\nWords without Punctuation:")
print("\nTokens without Punctuation")
print(words)

#step 4 : Convert to Lowercase
lower_words = [word.lower() for word in words]
print("\nLowercase Words:")
print(lower_words)

#step 5 : count tokens
print("\nTotal tokens : " ,len(tokens))
print("total words: ", len(words))
```

```python
#Create and writing to text file

# Open file in write mode
file= open("students.txt","w")

file.write("Rahul\n")
file.write("Priya\n")
file.write("Amit\n")
file.write("Neha\n")

file.close()
print("File created sccessfully.")
```

```python
# 2. Writing Multiple Lines Using writelines()

students = [
    "Rahul\n",
    "Priya\n",
    "Amit\n",
    "Neha\n"
]

with open ("students.txt", "w") as file:
  file.writelines(students)

print("Data written successfully")
```

```python
# Reading the entire file
with open("students.txt","r") as file:
  data=file.read()
print(data)
```

```python
#Reading file line by line
with open("students.txt","r") as file:
  for line in file:
    print(line.strip())   #Strip() removes the newline character (/n) from each line
```

```python
#using readline()
with open ("students.txt","r") as file:
  line1=file.readline()
  line2=file.readline()
print("First line: " , line1.strip())
print("Second line: ",line2.strip())
```

```python
#using readlines()
with open ("students.txt","r") as file:
  lines=file.readlines()
print(lines)
```

```python
#You can also process the list:
with open("students.txt","r") as file:
  lines=file.readlines()
  for line in lines:
    print(line.strip())
```

```python
# Appending Data to a file
with open("students.txt","a") as file:
  file.write("karan\n")
  file.write("Simran\n")
print("New Students added.")
```

```python
#writing user input to a file (Integers)
name=input("enter your name :")
age=int(input("Enter your age :"))

with open("user_data.txt","w") as file:
    file.write(f"Name: {name}\n")
    file.write(f"Age: {age}\n")

print("Information Saved.")
```

```python
#Counting Number of lines
with open("students.txt","r") as file:
  lines=file.readlines()
  line_count=len(lines)
print("Total number of lines:",line_count)
```

```python
# counting words in a text file
with open("students.txt","r") as file:
  data=file.read()

words=data.split()
print("Total number of words:",len(words))
```

```python
#counting character
with open("students.txt","r") as file:
  data=file.read()

char_count=len(data)
print("Total number of characters:",char_count)
```

```python
#exclude spaces
characters = data.replace(" ","").replace("\n","")
print("Characters excluding spaces:", len(characters))
```

```python
#searching for a word
search_word = input("Enter word to search: ")
with open("students.txt","r") as file:
  data=file.read()
  if search_word.lower() in data.lower():
    print("word found")
  else:
    print("not found")
```

```python
#counting occurrwnces of a word
Search_word = input("Enter word: ")
with open("students.txt","r") as file:
  data=file.read()
count = data.lower().count(search_word.lower())
print("occurrences of word:",count)
```

```python
#copying contents from one file to another
with open("students.txt","r") as source:
  data = source.read()
with open("sample.txt","w") as destination:
  destination.write(data)
print("File copied successfully.")
```

```python
import spacy
nlp=spacy.load("en_core_web_sm")
```

```python
#1 : check a word is a stopword
words = ["the","is","python","and","beautiful"]
for word in words:
  doc = nlp(word)
  for token in doc:
    print(token.text,"->",token.is_stop)
```

```python
#2 : Identify Stopwords in a sentence
text = "Python is a very useful programming language"
doc = nlp(text)
for token in doc:
  print(token.text,"->",token.is_stop)
```

```python
#3 : Extraxt only stopwords
text = "This is a simple example of stopwords removal"
doc = nlp(text)
stopwords = []
for token in doc:
  if token.is_stop:
    stopwords.append(token.text)
print(stopwords)
```

```python
#4 extrat non-stopwords
text = "Python is a very usefull and powerful language "
doc = nlp(text)
words = []
for token in doc:
  if not token.is_stop and not token.is_punct:
    words.append(token.text)
print(words)
```

```python
#5 : Remove Stopwords from a sentence
text = "Python is a very usefull and powerful language "
doc = nlp(text)
result = []
for token in doc:
  if not token.is_stop:
    result.append(token.text)
print(" ".join(result))
```

```python
#6: remove stopwords and punctuation
text = "The students are learning data sacience i the classroom"
doc=nlp(text)
words = []
for token in doc:
  if not token.is_stop and not token.is_punct:
    words.append(token.text)
print(" ".join(words))
```

```python
#7 : Convert text to lowercase and remove stopwaords
text = "Python is a very powerful programming language"
doc = nlp(text)
result = []
for token in doc:
  if not token.is_stop:
    result.append(token.text.lower())
print(" ".join(result))
```

```python
#10 : Create a stopword removal function
def remove_stopwords(text):
  doc = nlp(text)
  result=[]
  for token in doc :
    if not token.is_stop and not token.is_punct:
      result.append(token.text)
  return " ".join(result)

text = "Python is a very powerful programming language"
print(remove_stopwords(text))
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))
```

```python
#2 : Check Whether a word is a stopword
words = ["the","is","python","and","beautiful"]
for word in words:
 print(word,"->",word.lower() in stop_words)
```

```python
#3 : Extract Stopwords from a sentence
from nltk.tokenize import word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
text = "python is a very powerful programming language"
tokens = word_tokenize(text)
result = []
for word in tokens:
  if word.lower() in stop_words:
    result.append(word)
print(result)
```

```python
#4 : remove stopwords from a sentence
text = "The students are learning data science."
tokens = word_tokenize(text)
result = []
for word in tokens:
  if word.lower() not in stop_words:
    result.append(word)
print(result)
```

```python
#5: remove stopwords and punctuation
import string
text = "The students are learning data science!"
tokens = word_tokenize(text)
result = []
for word in tokens:
  if word.lower() not in stop_words and word not in string.punctuation:
    result.append(word)
print(result)
```

```python
#6: convert text to lowercase and remove stopwords
text = "Python is a powerful programming language."
tokens = word_tokenize(text)
result = []
for word in tokens:
  if word.lower() not in stop_words:
    result.append(word.lower())
print(result)
```

```python
#7: count stopwords in a text
text = "This is a simple example to demonstrate stopword removal "
tokens = word_tokenize(text)
count = 0
for word in tokens:
  if word.lower() in stop_words:
    count += 1
print("Number of stop_words:- ",count)
```

```python
#9 Add custom stopwords in NLTK
custom_words = {"Python","data","language"}
stop_words.update(custom_words)
text = "Python is a data science language"
tokens = word_tokenize(text)
result = []
for word in tokens:
  if word.lower() not in stop_words:
    result.append(word)
print(result)
```

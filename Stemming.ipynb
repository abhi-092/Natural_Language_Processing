###Stemming in NLP###

Stemming is the process of reducing words to their root/base form (stems) by removing preffixes or suffixes.



Important : spaCy does not provide a built-in stemming algorithm. spaCy primarly uses lemmatization. If you specifically want stemming while using spaCy , you can integrate a stemmer such as NLTK's porterstemmer or snowballStemmer.

***stemming using NLTK***

```python
#import poterStemmer
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
```

```python
#1 : Basic stemming
words = ["playing","played","plays","player"]
for word in words:
  print(word,"->",stemmer.stem(word))
```

```python
#2 : Stemming a sentence
text = "The students are studying different subjects."
words = text.split()
for word in words:
  print(word,"->",stemmer.stem(word))
words =["Playing","played","plays","player"]
for word in words:
  print(word,"->",stemmer.stem(word))
```

```python
#3 : words ending with ing
words = [
    "playing",
    "reading",
    "writing",
    "walking",
    "running"
]
for word in words:
  print(word,"->",stemmer.stem(word))
```

```python
#4 : words ending with "ed"
words = [
    "played",
    "walked",
    "talked",
    "connected",
    "completed"
]
for word in words:
  print(word,"->",stemmer.stem(word))
```

```python
#5 : words ending with "s"
words= [
    "cats",
    "dogs",
    "books",
    "students",
    "cars"
]
for word in words:
  print(word,"->",stemmer.stem(word))
```

```python
#6 : different forms of the same word
words = [
    "connect",
    "connected",
    "connecting",
    "connection",
    "connections"
]
for word in words:
  print(word,"->",stemmer.stem(word))
```

```python
#7 : Stemming with a loop
words = [
    "happy",
    "happier",
    "happiest",
    "happying",
    "happiness"
]
for word in words:
  print(word,"->",stemmer.stem(word))
```

```python
#8 : Stemming words related to organize
words = [
    "organize",
    "organized",
    "organizing",
    "organization",
    "organizations"
]
for word in words:
  print(word,"->",stemmer.stem(word))
```

```python
#9 : Stemming text using regular expressions
import re
text = "Students are learning Python and developing machine learning projects"
words = re.findall(r'\b[a-zA-Z]+\b',text.lower())
for word in words:
  print(word,"->",stemmer.stem(word))
```

```python
#10 : create a stemmed sentence
text = "The players are playing football and practicing regularly"
words = text.lower().split()
stemmed_words = []
for word in words:
  stemmed_words.append(stemmer.stem(word))
stemmed_text = " ".join(stemmed_words)
print("Original:")
print(text)
print("\nStemmed:")
print(stemmed_text)
```

Stemming with spaCy

Impotant note:
spaCy does not have a native stem()function like NLTK
Therefore , if we want to demonstrate stemming in a spaCy work flow we can use: spaCy >>>> tokenization >>>>> NLTK stemmer

```python
!pip install spacy nltk
```

```python
!python -m spacy download en_core_web_sm
```

```python
# importing library
import spacy
from nltk.stem import PorterStemmer

nlp = spacy.load("en_core_web_sm")
stemmer = PorterStemmer()
```

```python
#1 : Stemming spacy tokens
text = "the students are studying."
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text,"->",stemmer.stem(token.text))
```

```python
#2: playing , played ,plays
text = "I Play football. He plays football. They played yesterday"
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text,"->",stemmer.stem(token.text))
```

```python
#3 : reading related words
text = "Reading readers read readable books"
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text,"->",stemmer.stem(token.text))
```

```python
#4: Learning related words
text = "Students learn learning learned concepts"
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text,"->",stemmer.stem(token.text))
```

```python
#5 : connect related words
text = "the system connects connected connecting devices"
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text,"->",stemmer.stem(token.text))
```

```python
#6 : Stemming a complete paragraph
text = """
Data scientists analyze data.
They are analyzing datasets and developed useful models.
"""
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(f"{token.text:15}->{stemmer.stem(token.text)}")
```

```python
#7 : removing punctuation before stemming
text = "Python is powerfull , usefull , and popular"
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text,"->",stemmer.stem(token.text))
```

```python
#8 : stemming with lowercase conversion
text = "Python Pythonic Programming programs"
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    word = token.text.lower()
    print(word,"->",stemmer.stem(word))
```

```python
#9: Create a stemmed sentence using spacy
text = "The children are playing games and enjoying themselves"
doc = nlp(text)
stemmed_words = []
for token in doc:
  if token.is_alpha:
    stemmed_words.append(stemmer.stem(token.text.lower()))

stemmed_text = " ".join(stemmed_words)
print("Original:")
print(text)
print("\nStemmed:")
print(stemmed_text)
```

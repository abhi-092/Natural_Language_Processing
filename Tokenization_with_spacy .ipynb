```python
#then import spacy
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
text = "python is easy to learn"
doc=nlp(text)
for token in doc:
  print(token.text)
```

```python
# tokenization with token index
#Each token has index indicating its position in the document
# token.i gives the position/index of the token
text = "Python is easy to learn."
doc = nlp(text)

for token in doc:
  print(token.i,token.text)
```

```python
#tokenization with token attributes
nlp = spacy.load('en_core_web_sm')
text = "Python is easy to learn."
doc = nlp(text)

for token in doc:
  print(token,token.text,token.is_alpha,token.is_punct,token.is_space)
```

```python
# tokenization of a sentence with punctuation
text = "Hello, how are you?"
doc = nlp(text)

for token in doc:
  print(token.text)
```

```python
#tokenization of number
text = "I brought 5 books $100"
doc = nlp(text)

for token in doc:
  print(token.text)
```

```python
#tokenization of email address
text = "Contact me at neem@example.com"
doc = nlp(text)

for token in doc:
  print(token.text)
```

```python
#tokenization of urls
text = "visit my website at https://www.google.com"
doc = nlp(text)

for token in doc:
  print(token.text)
```

```python
#tokenization of contractions
text = "i'm going to park"
doc = nlp(text)

for token in doc:
  print(token.text)
```

```python
#tokenization of hyphenated word
text = "Data-driven decision-making is importtant"
doc = nlp(text)

for token in doc:
  print(token.text)
```

```python
#converting token into a list
text = "Python is easy to learn."
doc = nlp(text)
token_list = [token.text for token in doc]
print(token_list)
```

```python
# Removing punctuation after tokenization
text ="python is powerful, easy, and popular!"
doc = nlp(text)
for token in doc:
  if not token.is_punct:
    print(token.text)
```

```python
# Getting only alphabetic token
# we can use is_alpha to select only alphabetic token.
text ="python 3 is easy to learn!"
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text)
```

```python
#tokenization of multiple sentence
text = "Python is easy to learn. Python is powerful."
doc = nlp(text)

for token in doc:
  print(token.text)

```

```python
#tokenization and sentence segmentation
#spacy can also identify individual sentences
text = "Python is easy to learn. Python is powerful."
doc = nlp(text)

for sent in doc.sents:
  print("sentence : ",sent.text)
  for token in sent:
    print(token.text)

  print()
```

```python
#tokenization from a text file
nlp = spacy.load("en_core_web_sm")
with open("/content/11111.txt","r") as file:
  text = file.read()
doc = nlp(text)

for token in doc:
  print(token.text)
```

```python
#complete example
text = "Python is easy to learn. Python is powerful."
doc = nlp(text)

print("Token\tAlpha\tNumber\tPunctuation\tSpace")
for token in doc:
  print(token.text,"\t",token.is_alpha,"\t",token.is_digit,"\t",token.is_punct,"\t",token.is_space)
```

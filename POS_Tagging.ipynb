What is POS tagging?

POS tagging (Part-of-speech) is the process of assigning a grammatical category to each word in sentence.

```python
#2 : install spacy
import spacy
```

```python
#Download the english model:
!python -m spacy download en_core_web_sm
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
#1:Basic POS tagging
text = "python is usefull."
doc = nlp(text)
for token in doc:
  print(token.text,token.pos_,token.pos)
```

```python
#2: Display word , POS and Explanation
text = "The students writes python code"
doc = nlp(text)
for token in doc:
  print(token.text,
        token.pos_,
        spacy.explain(token.pos_)
        )
```

```python
#3 : Nouns and verb
#we can extract only nouns and verb
text = "Students learn python and develop applications."
doc = nlp(text)
for token in doc:
  if token.pos_ in ['NOUN','VERB']:
    print(token.text,token.pos_)
```

```python
#4 : Extract Adjectives
text = "Python is a powerful and popular programming language."
doc = nlp(text)
for token in doc:
  if token.pos_ == 'ADJ':
    print(token.text,token.pos_)
```

```python
#5 : Extract adverbs
text = "The students quickly completed the assignment"
doc = nlp(text)
for token in doc:
  if token.pos_ == 'ADV':
    print(token.text,token.pos_)
```

```python
#6 : Extract proper noun
text = "Neema teaches Data Science at Delhi University."
doc = nlp(text)
for token in doc:
  if token.pos_=="PROPN":
    print(token.text,token.pos_)
```

```python
#7 : POS tagging with numbers
text = "There are 50 students in the class."
doc = nlp(text)
for token in doc:
  print(token.text,token.pos_)
```

```python
#8 : POS frequency
#we can count how many times each POS category occurs.
from collections import Counter
text = """
Python is a popular programming language.
Students use Python to develop machine learning applications.
Machine learning is useful in data science"""
doc = nlp(text)
pos_counts = Counter([token.pos_ for token in doc])
print(pos_counts)
```

```python
#9 : POS tagging in a sentence
text = "The intelligent student solved the difficult problem quickly"
doc = nlp(text)
for token in doc :
  print(f"{token.text:12}{token.pos_}")
```

```python
#10: POS Tagging with tag_
text = "She is studying python"
doc = nlp(text)
for token in doc:
  print(token.text,token.pos_,token.tag_)
```

```python
#11: POS Tagging with visualization
#spacy provides a built-in visualization tool called displacy
from spacy import displacy
text = "The intelligent student quickly solved the difficult problem."
doc = nlp(text)
displacy.render(doc,style='dep',jupyter=True)
```

```python
#12: Visualize POS Tags as a bar Chart
#we can count POS tag and display their frequency.
from collections import Counter
import matplotlib.pyplot as plt
from spacy import displacy
text = """
Python is a powerful programming language.
Students use Python to develop machine learning applications.
Machine learning is widely used in data science.
"""
doc = nlp(text)
pos_counts = Counter([token.pos_ for token in doc])
plt.figure(figsize=(10,5))
plt.bar(pos_counts.keys(),pos_counts.values())
plt.xlabel('POS Tags')
plt.ylabel('Frequency')
plt.title('POS Tag Frequency')
plt.show()
```

```python
#13: POS Visualization Using displacy
#spacy can visualize part-of-speech information directly.
text = "Python is a powerful programming language"
doc = nlp(text)
options={
    'compact':True,'distance':100
}
displacy.render(doc,style='dep',options=options,jupyter=True)
```
